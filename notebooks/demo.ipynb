# Demo — Agente Text-to-SQL

Este notebook muestra el agente en acción: recibe preguntas en lenguaje natural sobre datos de una distribuidora editorial y las traduce a SQL.

**Requisitos:** tener `data.db` creada (`py src/setup_db.py`) y `ANTHROPIC_API_KEY` en `.env`.

```python
import sys
sys.path.insert(0, '..')

from src.agent import AgenteSQL

agente = AgenteSQL()
```

```python
def mostrar(pregunta):
    """Ejecuta una pregunta y muestra SQL + resultado + explicación."""
    r = agente.consultar(pregunta)
    print(f"Pregunta: {pregunta}")
    print(f"\nSQL generado:\n{r['sql']}\n")
    display(r['resultado'].head(15))
    if r['explicacion']:
        print(f"\n💡 {r['explicacion']}")
```

## Ejemplo 1 — Top sucursales por ventas

```python
mostrar("¿Cuáles fueron las 5 sucursales con más ventas?")
```

## Ejemplo 2 — Riesgo de quiebre de stock

```python
mostrar("¿Qué títulos tienen ventas mayores al stock disponible?")
```

## Ejemplo 3 — Peso de categoría Juvenil

```python
mostrar("¿Cuánto representa la categoría juvenil sobre el total de unidades vendidas?")
```

## Ejemplo 4 — Top 10 títulos más vendidos

```python
mostrar("Mostrame el top 10 de títulos más vendidos")
```
